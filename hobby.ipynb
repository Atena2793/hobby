```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import LabelEncoder, OrdinalEncoder
from sklearn.linear_model import LogisticRegression
from sklearn.neighbors import KNeighborsClassifier
from sklearn.svm import SVC
from sklearn.tree import DecisionTreeClassifier
from sklearn.metrics import accuracy_score
```

```python
df = pd.read_csv(r"A:\practice_files\Hobby_Data.csv")
df.info()
df.isnull().sum()
df.duplicated().sum()
```

```python
df.isnull().sum()
```

```python
df = df.drop_duplicates()
```

```python
x = df.drop("Predicted Hobby", axis=1)
y = df["Predicted Hobby"]
```

```python
x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.2)
```

```python
#Predicted Hobby -> label
#won_arts -> ordinal encoder
#remainder -> get_dummies
le = LabelEncoder()
y_train = le.fit_transform(y_train)
y_test = le.transform(y_test)

oe = OrdinalEncoder(categories=[["Yes","Maybe","No"]])
x_train["Won_arts"] = oe.fit_transform(x_train[["Won_arts"]])
x_test["Won_arts"] = oe.transform(x_test[["Won_arts"]])

x_train = pd.get_dummies(x_train)
x_test = pd.get_dummies(x_test)
```

```python
lr = LogisticRegression().fit(x_train, y_train)
pred1 = lr.predict(x_test)
knn = KNeighborsClassifier().fit(x_train, y_train)
pred2 = knn.predict(x_test)
svm = SVC().fit(x_train, y_train)
pred3 = svm.predict(x_test)
tree = DecisionTreeClassifier().fit(x_train, y_train)
pred4 = tree.predict(x_test)
```

```python
print(f"LogisticRegression is: %{accuracy_score(y_test, pred1) * 100}")
print(f"KNeighborsClassifier is: %{accuracy_score(y_test, pred2) * 100}")
print(f"SVC is: %{accuracy_score(y_test, pred3) * 100}")
print(f"DecisionTreeClassifier is: %{accuracy_score(y_test, pred4) * 100}")
```
